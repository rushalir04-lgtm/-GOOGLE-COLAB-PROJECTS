```python
# NUMPY & PANDAS
import pandas as pd
import numpy as np

# sample numerical data
data = {
    "Age":[18 ,25,30,45,60],
    "Salary":[20000,30000,50000,80000,150000]
}
df = pd.DataFrame(data)
print(df)
```

```python
#  Feature Scaling Standardization --

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_scaled =df.copy()
df_scaled["Age"] = scaler.fit_transform(df_scaled[["Age"]])
df_scaled["Salary"] = scaler.fit_transform(df_scaled[["Salary"]])

print(df_scaled)
```

```python
#  Feature Scaling  Normalization -- Min & Max
from sklearn.preprocessing import MinMaxScaler

scalar = MinMaxScaler()

df_scaled =df.copy()
df_scaled["Age"] = scalar.fit_transform(df_scaled[["Age"]])
df_scaled["Salary"] = scalar.fit_transform(df_scaled[["Salary"]])
print(df_scaled)
```

```python
# Feature Scaling Log Transformation
# Apply Log Transformation
df_log = df.copy()
df_log["Age"] = np.log1p(df_log["Age"])
df_log["Salary"] = np.log1p(df_log["Salary"])
print(df_log)
```

```python
# Binning (Discreatization)
# convert to categories
bins =[0,20,40,100]
labels =["Young","Adult","Senior"]

df["Age_Group"]= pd.cut(df["Age"], bins=bins, labels =labels )
print(df)
```

```python
# Mean imputation
# Create a sample DataFrame with missing values to demonstrate mean imputation
data_missing = {
    'Feature_A': [10, 20, np.nan, 40, 50],
    'Feature_B': [1, np.nan, 3, 4, 5],
    'Feature_C': [100, 200, 300, np.nan, 500]
}
df_missing = pd.DataFrame(data_missing)

print("Original DataFrame with missing values:")
print(df_missing)
```

```python
# Perform mean imputation on the DataFrame
df_imputed_mean = df_missing.copy()
for column in df_imputed_mean.columns:
    if df_imputed_mean[column].isnull().any():
        mean_value = df_imputed_mean[column].mean()
        df_imputed_mean[column] = df_imputed_mean[column].fillna(mean_value)

print("DataFrame after mean imputation:")
print(df_imputed_mean)
```

```python
# Meadian imputation
df_imputed_median = df_missing.copy()
for column in df_imputed_median.columns:
    if df_imputed_median[column].isnull().any():
        median_value = df_imputed_median[column].median()
        df_imputed_median[column] = df_imputed_median[column].fillna(median_value)

print("DataFrame after median imputation:")
print(df_imputed_median)
```

```python
# Mode Imputatiuon
df_imputed_mode = df_missing.copy()
for column in df_imputed_mode.columns:
    if df_imputed_mode[column].isnull().any():
        # For numerical data, mode() can return multiple values if there's a tie.
        # We'll take the first mode if multiple exist.
        mode_value = df_imputed_mode[column].mode()[0]
        df_imputed_mode[column] = df_imputed_mode[column].fillna(mode_value)

print("DataFrame after mode imputation:")
print(df_imputed_mode)
```

```python
# Forward Fill
df_imputed_ffill = df_missing.copy()
df_imputed_ffill = df_imputed_ffill.ffill()

print("DataFrame after forward fill imputation:")
print(df_imputed_ffill)
```

```python
# Backword Fill
df_imputed_bfill = df_missing.copy()
df_imputed_bfill = df_imputed_bfill.bfill()

print("DataFrame after backward fill imputation:")
print(df_imputed_bfill)
```

```python
# Missing Indicator
# Create a copy to avoid modifying the original DataFrame
df_missing_indicator = df_missing.copy()

# Create missing indicator columns
for column in df_missing_indicator.columns:
    if df_missing_indicator[column].isnull().any():
        df_missing_indicator[f'{column}_is_missing'] = df_missing_indicator[column].isnull().astype(int)

print("DataFrame with missing indicators:")
print(df_missing_indicator)
```

## Feature Selection

Feature selection is the process of selecting a subset of relevant features (variables, predictors) for use in model construction. This helps to reduce dimensionality, improve model performance, and reduce overfitting.

### 1. Variance Thresholding

Removes features with low variance. It's a simple baseline approach for feature selection. Features with zero variance (i.e., constant values) provide no information and are removed. Features with very low variance might also be uninformative.

```python
from sklearn.feature_selection import VarianceThreshold

# Create a sample DataFrame with some low variance features
data_variance = {
    'Feature_A': [1, 2, 3, 4, 5],
    'Feature_B': [10, 10, 10, 10, 10], # Zero variance
    'Feature_C': [0.1, 0.2, 0.1, 0.2, 0.1], # Low variance
    'Feature_D': [100, 200, 300, 400, 500]
}
df_variance = pd.DataFrame(data_variance)

print("Original DataFrame for Variance Thresholding:")
print(df_variance)

# Initialize VarianceThreshold with a threshold (e.g., 0.1)
# Features with variance lower than this threshold will be removed
selector = VarianceThreshold(threshold=0.1)

# Fit and transform the data
df_variance_selected = selector.fit_transform(df_variance)

# Get the names of the selected features
selected_features = df_variance.columns[selector.get_support(indices=True)]

print(f"\nFeatures selected (threshold=0.1): {list(selected_features)}")
print("DataFrame after Variance Thresholding:")
print(pd.DataFrame(df_variance_selected, columns=selected_features))
```

### 2. Correlation-based Feature Selection

This method involves analyzing the correlation between features and either the target variable (for supervised learning) or between features themselves (to reduce multicollinearity). Highly correlated features can provide redundant information.

#### a. Removing Highly Correlated Features (between features)

This technique identifies pairs of features that are highly correlated with each other and removes one of them, typically the one that is less relevant or has lower importance to the model. This helps in reducing multicollinearity.

```python
# Create a sample DataFrame with highly correlated features
data_corr = {
    'Feature_X': [1, 2, 3, 4, 5],
    'Feature_Y': [2, 4, 6, 8, 10], # Highly correlated with Feature_X
    'Feature_Z': [5, 4, 3, 2, 1],  # Negatively correlated with Feature_X
    'Feature_W': [10, 11, 12, 13, 14] # Another feature
}
df_corr = pd.DataFrame(data_corr)

print("Original DataFrame for Correlation-based Feature Selection:")
print(df_corr)

# Calculate the correlation matrix
corr_matrix = df_corr.corr().abs()
print("\nCorrelation Matrix:")
print(corr_matrix)

# Select upper triangle of correlation matrix
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Find features with correlation greater than a threshold (e.g., 0.9)
to_drop = [column for column in upper_tri.columns if any(upper_tri[column] > 0.9)]

print(f"\nFeatures to drop due to high correlation (threshold > 0.9): {to_drop}")

# Drop features
df_corr_selected = df_corr.drop(columns=to_drop)

print("DataFrame after removing highly correlated features:")
print(df_corr_selected)
```

```python
# Correlation with Target Variable

# Create a sample DataFrame with features and a target variable
data_target_corr = {
    'Feature_1': [10, 20, 30, 40, 50],
    'Feature_2': [2, 4, 6, 8, 10], # Highly correlated with target
    'Feature_3': [100, 90, 80, 70, 60], # Negatively correlated with target
    'Feature_4': [5, 5, 6, 6, 7], # Low correlation
    'Target': [20, 40, 60, 80, 100]
}
df_target_corr = pd.DataFrame(data_target_corr)

print("Original DataFrame with Features and Target:")
print(df_target_corr)

# Calculate the correlation of each feature with the target variable
correlations = df_target_corr.corr()['Target'].abs().sort_values(ascending=False)
print("\nCorrelation of Features with Target (Absolute Values):")
print(correlations)

# Define a correlation threshold (e.g., 0.5)
correlation_threshold = 0.5

# Select features whose absolute correlation with the target is above the threshold
selected_features_with_target = correlations[correlations > correlation_threshold].index.tolist()

# Remove the 'Target' itself from the list of selected features if it's there
if 'Target' in selected_features_with_target:
    selected_features_with_target.remove('Target')

print(f"\nFeatures selected based on correlation with Target (threshold > {correlation_threshold}): {selected_features_with_target}")

# Display the DataFrame with only the selected features and the target
df_target_corr_selected = df_target_corr[selected_features_with_target + ['Target']]
print("DataFrame after selecting features based on correlation with Target:")
print(df_target_corr_selected)
```

Handling Outlier

```python
# Handling Outliers - IQR Method

# Create a sample DataFrame with some outliers
data_outliers = {
    'Feature_A': [10, 20, 30, 40, 50, 100, 10],
    'Feature_B': [1, 2, 3, 4, 5, 20, 2],
    'Feature_C': [100, 200, 300, 400, 500, 1500, 100]
}
df_outliers = pd.DataFrame(data_outliers)

print("Original DataFrame with potential outliers:")
print(df_outliers)

# Function to detect and cap outliers using IQR method
def cap_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap outliers
    df[column] = np.where(df[column] < lower_bound, lower_bound, df[column])
    df[column] = np.where(df[column] > upper_bound, upper_bound, df[column])
    return df

# Apply outlier capping to each numerical column
df_outliers_capped = df_outliers.copy()
for column in df_outliers_capped.columns:
    df_outliers_capped = cap_outliers_iqr(df_outliers_capped, column)

print("\nDataFrame after capping outliers (IQR method):")
print(df_outliers_capped)
```
