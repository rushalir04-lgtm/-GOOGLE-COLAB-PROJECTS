```python
class Dog:
  species="Canine"  #class Attribute

  def __init__(self, name, age):
    self.name= name   #instance variable
    self.age=age      #instance variable
  def animal_name(self, dog, cat):
    self.dog= dog
    self.cat=cat

#taking input from user
# For demonstration, I'm providing default values instead of interactive input
# name = input("enter dog object name :")
# age = int(input("enter dog object age :"))

name = "Max"
age = 5

# creating object
d=Dog(name ,age)
print(d.name)
print(d.age)
print(d.species)

# Defining dog_value and cat_value for demonstration
dog_value = "German Shepherd"
cat_value = "Persian"

#calling function with user input
d.animal_name(dog_value ,cat_value)
print(f"Dog type: {d.dog}, Cat type: {d.cat}")
```

```python
# TUPLE
# given a tuple of numbers and a target element, find and print all indexes where the target element occurs in the tuple.

my_tuple = (10, 20, 10, 30, 40, 10, 50, 20)
target_element = 10

print(f"Original tuple: {my_tuple}")
print(f"Target element: {target_element}")

indexes = []
for index, element in enumerate(my_tuple):
    if element == target_element:
        indexes.append(index)

if indexes:
    print(f"The target element {target_element} occurs at indexes: {indexes}")
else:
    print(f"The target element {target_element} was not found in the tuple.")
```

```python
# Union
set1={1,2,3,4}
set2={3,4,5,6}
result = set1.union(set2)
print("Union of set1 and set2:", result)
```

```python
# Intersection
set1={1,2,3,4}
set2={3,4,5,6}
result = set1.intersection(set2)
print("Intersection of set1 and set2:", result)
```
