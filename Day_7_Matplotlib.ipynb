```python
from matplotlib.lines import lineStyles
# importing Matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Sample Data
x=np.linspace(0,10, 10)
print(x)
y= np.sin(x)
# print(y)

# basic line plotting
plt.plot(x,y, label="sin(x)", color="Blue", linestyle="-." )
plt.xlabel('x_axis')
plt.ylabel('sin(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Example 1 : line plot with multiple lines
import numpy as np
# import matplotlib.pyplot as plt
y2= np.cos(x)
plt.plot(x,y, label="sin(x)", color="Blue", linestyle="-." )
plt.plot(x,y2, label="cos(x)", color="Red", linestyle="-." )
plt.xlabel('x_axis')
plt.ylabel('sin(x)')
plt.legend()
plt.title("sin(x) and cos(x) waves")
plt.grid(True)
plt.show()
```

```python
# Example 2 : Bar chart
categories = ['A','B','C','D']
values =[10,24,36,18]
plt.bar(categories, values, color=['blue','orange','green','red'], label="random")
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart')
plt.legend()
# plt.grid(True)
plt.show()
```

```python
# Example 3 : Histogram
# students =[]
data = np.random.randn(100)
print(data)
plt.hist(data, bins=100, color='purple',edgecolor='black', label="ramdom")
plt.title('Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
# Example 4 : Histogram of students marks

# Generate sample student marks (e.g., normally distributed around 70, std dev 10)
student_marks = np.random.normal(loc=70, scale=10, size=100)

# Ensure marks are within a reasonable range (0-100)
student_marks = np.clip(student_marks, 0, 100)

plt.hist(student_marks, bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.grid(True)
plt.show()
```

```python
# Example 5 : Pie Chart
sizes= [60,20,25,25]
labels= ["Apple", "Banana", "Mango", "Grapes"]
plt.pie(sizes,labels=labels, autopct='%1.1f%%', startangle=90, shadow= True)
plt.title('Pie Chart')
plt.show()
```

```python
#  Example 6 :Scatter PLot
categories = ['A','B','C','D','E','F']
values =[10,24,36,18,34,23]
plt.scatter(categories, values, color='darkgreen', label="categories")
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Scatter Chart')
plt.legend()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

tips = sns.load_dataset("tips")
tips.head()
plt.figure(figsize=(10,6))
sns.scatterplot(x="total_bill", y="tip", data=tips, hue="sex")
plt.title("Scatter Plot of Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
print(tips)
sns.set_theme()
```

```python
# Line plot total bill vs Size
plt.figure(figsize=(10,6))
sns.lineplot(x="size", y="total_bill", data=tips)
plt.title("Line Plot of Total Bill vs Size")
plt.xlabel("Size")
plt.ylabel("Total Bill")
```

```python
# bar  plot total bill vs Size
plt.figure(figsize=(10,6))
sns.barplot(x="size", y="total_bill", data=tips)
plt.title("Bar Plot of Total Bill vs Size")
plt.xlabel("Size")
plt.ylabel("Total Bill")
```

```python
# violin plot total bill vs Size
plt.figure(figsize=(10,6))
sns.violinplot(x="size", y="total_bill", data=tips)
plt.title("Violin Plot of Total Bill vs Size")
plt.xlabel("Size")
plt.ylabel("Total Bill")
```

```python
# count  plot total bill vs Size
plt.figure(figsize=(10,6))
sns.countplot(x="size", data=tips)
plt.title("Count Plot of Size")
plt.xlabel("Size")
plt.ylabel("Count")
```

```python
# pie plot of Size distribution
plt.figure(figsize=(10,6))
size_counts = tips['size'].value_counts()
plt.pie(size_counts.values, labels=size_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Pie Plot of Size Distribution")
# plt.xlabel("Size") # Not applicable for pie charts
# plt.ylabel("Count") # Not applicable for pie charts
plt.show()
```

```python
# Box plot total bill vs Size
plt.figure(figsize=(10,6))
sns.boxplot(x="size", y="total_bill", data=tips)
plt.title("Box Plot of Total Bill vs Size")
plt.xlabel("Size")
plt.ylabel("Total Bill")
```

```python
# histogram  plot total bill vs Size
plt.figure(figsize=(10,6))
sns.histplot(x="total_bill", data=tips, kde=True)
plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
```

```python
# Heatmap plot of numerical correlations
plt.figure(figsize=(10,6))
correlation_matrix = tips.select_dtypes(include=['number']).corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title("Heatmap of Numerical Correlation Matrix")
plt.show()
```

```python
# violin plot distribution by gender and bill vs male female
plt.figure(figsize=(10,6))
sns.violinplot(x="sex", y="total_bill", data=tips)
plt.title("Violin Plot of Total Bill by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Bill")
```

```python
# pairplot total bill vs Size
sns.pairplot(tips)
plt.title("Pair Plot of Total Bill vs Size")
plt.show()
```

```python
# Regression plot total bill vs Size
sns.regplot(x="total_bill", y="size", data=tips)
plt.title("Regression Plot of Total Bill vs Size")
plt.xlabel("Total Bill")
plt.ylabel("Size")
plt.show()
```

```python
# joint plot total bill vs Size
sns.jointplot(x="total_bill", y="size", data=tips)
plt.title("Joint Plot of Total Bill vs Size")
plt.show()
```

```python
# Distrributin plot total bill vs Size(Histogram vs KDE)
sns.displot(x="total_bill", data=tips, kde=True)
plt.title("Distribution Plot of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.show()
```

```python
# Regression plot tips vs total bill
sns.regplot(x="total_bill", y="tip", data=tips)
plt.title("Regression Plot of Tip vs Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```
